```python
import requests

def get_news_about(company_name):

    url = ('https://newsapi.org/v2/everything?'
        f'q={company_name}&'
        'from=2024-09-20&'
        'apiKey=6284d539d651404fad9168a3d36ec004')

    response = requests.get(url)
    return response.json()

print(get_news_about('Akdital'))
```

```python
from newsapi import NewsApiClient

# Init
newsapi = NewsApiClient(api_key='6284d539d651404fad9168a3d36ec004')

# /v2/top-headlines
top_headlines = newsapi.get_top_headlines(q='Akdital',
                                          language='fr',
                                          country='ma')

# /v2/everything
all_articles = newsapi.get_everything(q='Akdital',
                                      from_param='2024-09-01',
                                      to='2024-09-29',
                                      language='fr',
                                    sort_by='relevancy',
                                    page=2)
```
